## 1.1 Latar Belakang
Python merupakan salah satu bahasa pemrograman yang banyak digunakan karena memiliki sintaks yang sederhana dan mudah dipahami terutama bagi pemula. Dalam mempelajari pemrograman Python, terdapat beberapa konsep dasar yang perlu dipahami sebelum membuat program yang lebih kompleks, salah satunya adalah variabel dan tipe data.

Variabel merupakan tempat untuk menyimpan suatu nilai atau data yang dapat digunakan kembali dalam program. Dengan menggunakan variabel, data dapat disimpan, diubah, dan diproses sesuai dengan kebutuhan program. Sementara itu, tipe data digunakan untuk menentukan jenis data yang disimpan dalam sebuah variabel. Beberapa tipe data dasar yang terdapat pada Python antara lain integer untuk bilangan bulat, float untuk bilangan desimal, string untuk teks, dan boolean untuk nilai benar atau salah.

Pemahaman mengenai variabel dan tipe data sangat penting karena hampir setiap program membutuhkan proses penyimpanan dan pengolahan data. Penggunaan tipe data yang sesuai juga dapat membantu program dalam melakukan operasi atau proses tertentu dengan benar. Oleh karena itu, pengenalan tipe data dan variabel menjadi salah satu dasar yang perlu dikuasai dalam pembelajaran pemrograman Python.

### 2.2.3 Program "Konstanta_2611531013.py"


```python
#Program ini menggunakan konstanta untuk menghitung luas lingkaran
#Nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14

print("pi: %f" % (PI))

jari_1013 = float(input('Masukkan nilai jari-jari: '))
luas_1013 = PI * jari_1013 * jari_1013

print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1013, luas_1013))
```
