# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori
Dasar teori pada praktikum ini membahas beberapa konsep yang berkaitan dengan bahasa pemrograman Python, Visual Studio Code (VS Code), dan GitHub. Ketiga hal tersebut merupakan bagian yang dapat digunakan dalam proses pembuatan dan pengelolaan suatu project pemrograman.

1. Bahasa Pemrograman Python


    Python merupakan bahasa pemrograman komputer yang biasa digunakan untuk membangun situs, aplikasi, mengotomatiskan tugas, dan melakukan analisis data. Bahasa pemrograman ini termasuk bahasa yang dapat digunakan untuk membuat berbagai jenis program dan tidak terbatas pada masalah tertentu saja. Karena sifatnya yang serbaguna dan mudah digunakan, Python banyak digunakan, terutama oleh pemula yang baru belajar pemrograman. Nama Python sendiri berasal dari Monty Python. Ketika Guido van Rossum membuatnya, ia sedang membaca skrip Sirkus Terbang Monty Python dari BBC. Menurutnya, nama tersebut terdengar singkat dan sedikit misterius, sehingga dipilih sebagai nama bahasa pemrograman yang dibuatnya.

2. VS Code


    VS Code merupakan code editor gratis yang dapat dijalankan pada perangkat desktop berbasis Linux, Windows, dan Mac OS yang dikembangkan oleh Microsoft. Aplikasi ini merupakan software editor yang powerful namun tetap ringan saat digunakan. VS Code dapat digunakan untuk membuat dan mengedit source code menggunakan berbagai bahasa pemrograman seperti Node.js, TypeScript, dan JavaScript, serta kompatibel dengan bahasa dan runtime environment lainnya seperti .NET, Java, Python, dan PHP.
    VS Code memiliki juga banyak fitur unggulan yang tidak dimiliki oleh beberapa pesaingnya sehingga membuat aplikasi ini banyak diminati oleh para pengguna.

3. GitHub


    GitHub merupakan platform berbasis web yang digunakan oleh developer untuk menyimpan, mengelola, dan mengembangkan proyek atau kode program secara bersama-sama. GitHub menyediakan layanan cloud untuk menyimpan repository sehingga pengguna dapat mengakses, mengedit, dan berkolaborasi dalam suatu proyek secara online. Selain sebagai tempat penyimpanan kode, GitHub juga berfungsi sebagai sistem pengelolaan versi kode dan platform kolaborasi bagi para developer dalam mengembangkan perangkat lunak.

### 2.2.2 Membuat Repository GitHub

Setelah program dibuat dan diuji pada VS Code, proyek dapat dihubungkan dengan GitHub. Repository digunakan untuk menyimpan kode program secara daring.

Tahapan yang dilakukan yaitu:

1. Membuat repository baru pada GitHub.
2. Membuka folder proyek menggunakan VS Code.
3. Membuka terminal pada VS Code.
4. Menginisialisasi Git pada folder proyek.
5. Menambahkan file ke dalam staging.
6. Membuat commit.
7. Menghubungkan repository lokal dengan repository GitHub.
8. Melakukan push untuk mengunggah kode ke GitHub

Contoh perintah yang digunakan:


```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

Penjelasan:

1. git init digunakan untuk membuat repository Git pada folder proyek.
2. git add . digunakan untuk menambahkan seluruh perubahan ke staging area.
3. git commit digunakan untuk menyimpan perubahan dengan sebuah pesan.
4. it branch -M main digunakan untuk menentukan branch utama dengan nama main.
5. git remote add origin digunakan untuk menghubungkan repository lokal dengan repository GitHub.
6. git push digunakan untuk mengunggah kode dari komputer ke GitHub.
