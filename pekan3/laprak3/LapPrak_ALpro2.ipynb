## 2.1 Dasar Teori
Dasar teori pada praktikum ini membahas beberapa konsep yang berkaitan dengan bahasa pemrograman Python, konsep variabel dalam python, dan tipe data. Ketiga hal tersebut merupakan bagian yang dapat digunakan dalam proses pembuatan dan pengelolaan suatu project pemrograman.

1. Bahasa Pemrograman Python

    Python merupakan bahasa pemrograman komputer yang digunakan untuk membangun situs, aplikasi, mengotomatiskan tugas, serta melakukan analisis dan visualisasi data. Python termasuk bahasa pemrograman tujuan umum sehingga dapat digunakan untuk berbagai kebutuhan dan tidak terbatas pada satu masalah tertentu. Karena mudah dipelajari dan serbaguna, Python banyak digunakan oleh pemula maupun profesional, bahkan oleh non-programmer seperti ilmuwan dan akuntan. Nama Python sendiri berasal dari acara BBC Monty Python’s Flying Circus yang sedang dibaca oleh Guido van Rossum saat mengembangkan bahasa pemrograman tersebut.

2. Konsep Variabel dalam Python

    Variabel dalam Python adalah nama yang digunakan untuk merujuk pada suatu objek atau nilai yang tersimpan di memori komputer. Variabel memudahkan programmer dalam menyimpan, menggunakan, dan memanipulasi data di dalam program. Variabel dibuat dengan memberikan nilai menggunakan sintaks variable_name = value, sehingga nilai tersebut dapat digunakan kembali dengan memanggil nama variabelnya.

3. Tipe Data

    Dalam Python, setiap data memiliki tipe yang menentukan bagaimana data tersebut disimpan dan operasi apa yang dapat dilakukan. Empat tipe data dasar yang dibahas adalah integer (int) untuk bilangan bulat, float (float) untuk bilangan desimal, string (str) untuk teks atau kumpulan karakter, dan boolean (bool) untuk nilai True atau False. Pemahaman terhadap tipe data ini penting sebagai dasar dalam membuat dan mengolah program Python. 
    

## 2.2 Kode Program

### 2.2.1 Program "Numeric_2611531013.py"


```python
#create a variable with integer value.
a_1013 = 100
print("The type of variable having value", a_1013, "is",
    type(a_1013))

#create a variable with float value.
b_1013 = 10.2345
print("The type of variable having value", b_1013, "is", 
    type (b_1013))

#create a variable with complex value 
c_1013 = 100 + 3j
print("The type of variable having value", c_1013, "is",
    type(c_1013))
```

**Penjelasan :**
1. **Membuat Komentar** :
    Sintaks #Create a variable with... digunakan untuk membuat komentar atau penanda yang menjelaskan fungsi algoritma yang ada di bawahnya. Komentar diawali dengan tanda # dan tidak akan dijalankan sebagai perintah oleh Python.

2. **Mendeklarasikan Variabel dan Nilainya**:
    Sintaks a_1013, b_1013, dan c_1013 merupakan variabel, sedangkan 100, 10.2345, dan 100 + 3j merupakan nilai yang disimpan ke dalam masing-masing variabel. Tanda = digunakan untuk memberikan nilai kepada variabel. Variabel berfungsi sebagai wadah penyimpanan nilai yang nantinya dapat diproses dalam program.

3. **Menentukan Tipe Data** :
    Sintaks type(a_1013), type(b_1013), dan type(c_1013) menggunakan fungsi type() untuk mengetahui tipe data dari nilai yang terdapat pada variabel. Tanda kurung () digunakan untuk memasukkan variabel atau nilai yang ingin diperiksa oleh fungsi type(). Output yang dihasilkan berupa <class 'data_type'>.
    
4. **Mengeluarkan Output Program** :
    Sintaks print("The type of variable having value", variable, "is", type(variable)) digunakan untuk menampilkan nilai variabel beserta tipe datanya ke layar. Fungsi print() digunakan untuk menghasilkan output, sedangkan tanda koma , digunakan untuk memisahkan beberapa data yang ingin ditampilkan.
    
5. **Tipe Data yang Digunakan** :
    Program menggunakan tiga tipe data numerik, yaitu int untuk bilangan bulat, float untuk bilangan desimal, dan complex untuk bilangan kompleks yang menggunakan j sebagai bagian imajinernya.



### 2.2.2 Program "String_2611531013.py"


```python
# Nama variabel ditambah 4 digit nim terakhir contoh: nama_1234
# Ganti programnya dan data masing masing

nama_1013 = "Fakhri Al Ghifari Busra"
jenis_kelamin_1013 = 'L'
alamat_1013 = """
    Koto Panjang
    Kecamatan Pauh
    Kota Padang
"""
agama_1013 = "Islam"

print("Nama saya adalah : ", nama_1013)
print("Jenis kelamin saya adalah : ", jenis_kelamin_1013)
print("Alamat saya adalah : ", alamat_1013)
print("Agama saya adalah : ", agama_1013)
```

3. **Menyimpan Data** :
Variabel nama_1013 menyimpan data nama, jenis_kelamin_1013 menyimpan jenis kelamin, alamat_1013 menyimpan alamat, sedangkan agama_1013 menyimpan data agama. Penambahan _1013 pada nama variabel digunakan sebagai identitas sesuai ketentuan penggunaan empat digit terakhir NIM.

4. **Menampilkan Data** :
Fungsi print() digunakan untuk menampilkan data yang telah disimpan dalam variabel ke layar. Tanda () digunakan untuk memasukkan data atau variabel yang ingin ditampilkan, sedangkan tanda koma , digunakan untuk memisahkan teks dengan variabel dalam output.

### 2.2.3 Program "Konstanta_2611531013.py"


```python
#Program ini menggunakan konstanta untuk menghitung luas lingkaran
#Nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14

print("pi: %f" % (PI))

jari_1013 = float(input('Masukkan nilai jari-jari: '))
luas_1013 = PI * jari_1013 * jari_1013

print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1013, luas_1013))
```

**Penjelasan :**
1. **from typing import Final** :
    Digunakan untuk mengimpor Final yang berfungsi membuat sebuah nilai sebagai konstanta.


2. **PI: Final = 3.14** :
    Mendefinisikan konstanta PI dengan nilai 3.14 yang digunakan dalam perhitungan luas lingkaran.

3. **print("pi: %f" % (PI))** :
    Menampilkan nilai PI ke layar dengan format angka desimal.

4. **jari_1013 = float(input('Masukkan nilai jari-jari: '))** :
    Meminta pengguna memasukkan nilai jari-jari, kemudian mengubah input tersebut menjadi tipe data float.

5. **luas_1013 = PI * jari_1013 * jari_1013** :
    Menghitung luas lingkaran menggunakan nilai PI dan jari-jari dengan rumus π × r × r.

6. **print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1013, luas_1013))** :
    Menampilkan hasil perhitungan luas lingkaran. %.2f digunakan agar angka ditampilkan dengan 2 angka di belakang koma.



### 2.2.4 Program "Input_2611531013.py"


```python
# Buat program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang 1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# Memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

panjang_1013 = input('Masukkan nilai panjang: ')
lebar_1013 = input('Masukkan nilai lebar: ')

luas_1013 = int(panjang_1013) * int(lebar_1013)
print("Luas = ", luas_1013)
```

**Penjelasan :**
1. **panjang_1013 = input('Masukkan nilai panjang: ')** :
    Digunakan untuk meminta pengguna memasukkan nilai panjang persegi panjang.

2. **lebar_1013 = input('Masukkan nilai lebar: ')** :
    Digunakan untuk meminta pengguna memasukkan nilai lebar persegi panjang.

3. **luas_1013 = int(panjang_1013) * int(lebar_1013)** :
    Menghitung luas persegi panjang dengan mengalikan panjang dan lebar. Nilai input diubah menjadi tipe data integer menggunakan int().
    
4. **print("Luas = ", luas_1013)** :
    Menampilkan hasil perhitungan luas persegi panjang ke layar.
