## 1.1 Latar Belakang
Dalam pemrograman, pengolahan data tidak hanya berkaitan dengan penyimpanan data ke dalam variabel, tetapi juga mencakup proses bagaimana data tersebut diolah dan dimanipulasi untuk mendapatkan hasil yang sesuai. Salah satu komponen dasar yang digunakan dalam proses tersebut adalah operator. Operator merupakan simbol atau instruksi khusus yang berfungsi untuk melakukan berbagai operasi terhadap nilai atau variabel (operand), seperti perhitungan matematika, perbandingan nilai, manipulasi bit, dan pengujian kondisi logika.

## 2.1 Dasar Teori
Operator adalah tanda atau kata tertentu dalam bahasa pemrograman yang digunakan untuk melakukan suatu tindakan terhadap nilai atau variabel (operand), sehingga menghasilkan suatu nilai atau hasil tertentu. Dalam Python, operator dapat dibedakan berdasarkan fungsi yang digunakan menjadi tujuh jenis utama.

1. Operator Aritmetika (Arithmetic Operators)

    Operator aritmetika digunakan ketika program perlu melakukan perhitungan terhadap bilangan. Beberapa operasi yang tersedia dalam Python meliputi penjumlahan ("+"), pengurangan ("-"), perkalian (""), pembagian ("/"), pembagian dengan hasil bilangan bulat ("//"), mencari sisa pembagian ("%"), dan menghitung pangkat ("*").

2. Operator Penugasan (Assignment Operators)

    Operator penugasan digunakan untuk menetapkan sebuah nilai ke dalam variabel. Bentuk paling sederhana adalah tanda sama dengan ("="). Python juga menyediakan operator penugasan yang dikombinasikan dengan operasi tertentu, seperti "+=", "-=", "*=", "/=", dan "%=", sehingga proses mengubah nilai variabel dapat ditulis dengan lebih ringkas.

3. Operator Perbandingan (Relational/Comparison Operators)

    Operator perbandingan digunakan ketika dua nilai atau variabel perlu diperiksa hubungannya. Perbandingan tersebut akan menghasilkan nilai Boolean berupa "True" atau "False". Jenis operator yang digunakan antara lain "==" untuk mengecek kesamaan, "!=" untuk ketidaksamaan, ">" untuk lebih besar, "<" untuk lebih kecil, ">=" untuk lebih besar atau sama dengan, serta "<=" untuk lebih kecil atau sama dengan.

4. Operator Logika (Logical Operators)

    Operator logika digunakan untuk menggabungkan beberapa kondisi atau menentukan hasil dari suatu pernyataan logis. Python memiliki tiga operator logika utama, yaitu "and", "or", dan "not". "and" menghasilkan "True" apabila kondisi yang diperiksa terpenuhi seluruhnya, "or" menghasilkan "True" apabila terdapat setidaknya satu kondisi yang benar, sedangkan "not" digunakan untuk menghasilkan kebalikan dari suatu nilai Boolean.

5. Operator Bitwise (Bitwise Operators)

    Operator bitwise digunakan untuk melakukan pengolahan terhadap bilangan berdasarkan representasi binernya, yaitu dalam bentuk bit "0" dan "1". Pengoperasian pada tingkat bit tersebut dapat dilakukan menggunakan beberapa operator, seperti AND ("&"), OR ("|"), XOR ("^"), NOT ("~"), pergeseran bit ke kiri ("<<"), dan pergeseran bit ke kanan (">>").

6. Operator Keanggotaan (Membership Operators)

    Operator keanggotaan digunakan untuk mengetahui apakah suatu elemen terdapat di dalam sebuah kumpulan atau struktur data. Pemeriksaan tersebut dapat dilakukan pada objek seperti string, list, tuple, maupun dictionary. Python menyediakan "in" untuk mengecek apakah suatu elemen terdapat di dalam objek dan "not in" untuk mengecek apakah elemen tersebut tidak terdapat di dalamnya.

7. Operator Identitas (Identity Operators)

    Operator identitas digunakan untuk mengetahui apakah dua variabel menunjuk atau merujuk pada objek yang sama. Python menyediakan "is" dan "is not" untuk melakukan pemeriksaan tersebut. Operator "is" menghasilkan "True" ketika kedua variabel merujuk pada objek yang sama, sedangkan "is not" menghasilkan "True" apabila keduanya tidak merujuk pada objek yang sama

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel di tambah 4 digit nim terakhir contoh : angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan di konversi menjadi tipe data integer

angka1_1013 = int(input("input angka-1: "))
angka2_1013 = int(input("input angka-2: "))

# Penjumlahan 
hasil_1013 = angka1_1013 + angka2_1013
print("\nOperator Penjumlahan")
print("Hasil =", hasil_1013)

# Pengurangan 
hasil_1013 = angka1_1013 - angka2_1013
print("\nOperator Pengurangan")
print("Hasil =", hasil_1013)

# Perkalian
hasil_1013 = angka1_1013 * angka2_1013
print("\nOperator Perkalian")
print("Hasil =", hasil_1013)

# Pembagian, Pembagian bilangan bulat, dan sisa bagi
if angka2_1013 != 0:
    hasil_1013 = angka1_1013 / angka2_1013
    print("\nOperator Pembagian")
    print("Hasil", hasil_1013)

    hasil_1013 = angka1_1013 // angka2_1013
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_1013)

    hasil_1013 = angka1_1013 % angka2_1013
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_1013)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_1013 = angka1_1013 ** angka2_1013
print("\nOperator Pangkat")
print("Hasil =", hasil_1013)

```

### 2.2.2 Program "assignment_2611531013.py"


```python
# Buat file dengan nama assignment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assignment dalam Python

angka1_1013 = int(input("Input angka-1: ")) 
angka2_1013 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =", angka1_1013)
print("Nilai angka2 =", angka2_1013)

# Assignment biasa
hasil_1013 = angka1_1013
print("\nAssignment biasa (=)")
print("Hasil =", hasil_1013)

# Assignment penambahan.
hasil_1013 = angka1_1013
hasil_1013 += angka2_1013
print("\nAssignment penambahan (+=)")
print("Hasil =", hasil_1013)

# Assignment pengurangan
hasil_1013 = angka1_1013
hasil_1013 -= angka2_1013
print("\nAssignment pengurangan (-=)")
print("Hasil =", hasil_1013)

# Assignment perkalian
hasil_1013 = angka1_1013
hasil_1013 *= angka2_1013
print("\nAssignment perkalian (*=)")
print("Hasil =", hasil_1013)

 # Assignment pembagian, pembagian bulat, dan sisa bagi
if angka2_1013 != 0:
    hasil_1013 = angka1_1013
    hasil_1013 /= angka2_1013
    print("\nAssignment pembagian (/=)")
    print("Hasil =", hasil_1013)  
    # Operator tambahan.
    hasil_1013 = angka1_1013
    hasil_1013 //= angka2_1013
    print("\nAssignment pembagian bulat (//=)")
    print("Hasil =", hasil_1013)
    hasil_1013 = angka1_1013
    hasil_1013 %= angka2_1013
    print("\nAssignment sisa bagi (%=)")
    print("Hasil =", hasil_1013)

else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assignment perpangkatan
hasil_1013 = angka1_1013 
hasil_1013 **= angka2_1013
print("\nAssignment perpangkatan (**=)")
print("Hasil =", hasil_1013)
```

Fungsi "input()" menerima data dalam bentuk string, kemudian "int()" mengubahnya menjadi tipe data integer agar dapat digunakan dalam operasi assignment.

3. **Menampilkan Nilai Awal** : 
    Perintah "print()" digunakan untuk menampilkan nilai awal dari variabel "angka1_1013" dan "angka2_1013" sebelum dilakukan proses assignment.

4. **Menerapkan Assignment Biasa** : 
    Sintaks "hasil_1013 = angka1_1013" menggunakan operator "=" untuk memberikan nilai dari "angka1_1013" ke variabel "hasil_1013". Nilai tersebut kemudian ditampilkan menggunakan "print()".

5. **Menerapkan Assignment Penambahan, Pengurangan, dan Perkalian** :

### 2.2.3 Program "bitwise_2611531013.py"


```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angkal_1234
# Program ini menggunakan fungsi input()

print("\n===================================")
print("3. OPERATOR BITWISE")
print("===================================") 

angka1_1013 = int(input("Masukkan angka bitwise-1: "))
angka2_1013 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =", angka1_1013, "| biner =", bin(angka1_1013))
print("angka2 =", angka2_1013, "| biner =", bin(angka2_1013))

#Bitwise AND
hasil_1013 = angka1_1013 & angka2_1013
print("\nBitwise AND (&)")
print(angka1_1013, "&", angka2_1013, "=", hasil_1013)
print("Biner hasil =", bin(hasil_1013))
print("Biner hasil (8 bit) =", format(hasil_1013, "08b"))

# Bitwise OR
hasil_1013 = angka1_1013 | angka2_1013
print("\nBitwise OR (|)")
print(angka1_1013, "|", angka2_1013, "=", hasil_1013)
print("Biner hasil =", bin(hasil_1013))
print("Biner hasil (8 bit) =", format(hasil_1013, "08b"))

# Bitwise XOR
hasil_1013 = angka1_1013 ^ angka2_1013
print("\nBitwise XOR (^)")
print(angka1_1013, "^", angka2_1013, "=", hasil_1013)
print("Biner hasil", bin(hasil_1013))
print("Biner hasil (8 bit)", format(hasil_1013, "08b"))

# Bitwise NOT
hasil_1013 = ~angka1_1013
print("\nBitwise NOT (~)")
print("~", angka1_1013, "=", hasil_1013)
print("Biner hasil =", bin(hasil_1013))
print("Biner hasil (8 bit) =", format(hasil_1013, "08b"))

# Bitwise geser kiri
jumlah_geser_1013 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_1013 = angka1_1013 << jumlah_geser_1013
print("\nBitwise geser kiri (<<)")
print(angka1_1013, "<<", jumlah_geser_1013, "=", hasil_1013)
print("Biner hasil =", bin(hasil_1013))
print("Biner hasil (8 bit) =", format(hasil_1013, "08b"))

# Bitwise geser kanan
hasil_1013 = angka1_1013 >> jumlah_geser_1013
print("\nBitwise geser kanan (>>)")
print(angka1_1013, ">>", jumlah_geser_1013, "=", hasil_1013)
print("Biner hasil =", bin(hasil_1013))
print("Biner hasil (8 bit) =", format (hasil_1013, "08b"))
```

### 2.2.4 Program "lainnya_2611531013.py"


```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("==================================")
print("1. OPERATOR KEANGGOTAAN")
print("==================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_1013 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_1013 = [int(angka.strip()) for angka in input_data_1013.split(",")]

nilai_dicari_1013 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_1013 = nilai_dicari_1013 in data_1013
print("\nOperator keanggotaan IN")
print(nilai_dicari_1013, "in", data_1013, "=", hasil_1013)

# Operator not in
hasil_1013 = nilai_dicari_1013 not in data_1013
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_1013, "not in", data_1013, "=", hasil_1013)

print("\n================================")
print("2. OPERATOR IDENTITAS")
print("==================================")

# objek1 menggunakan list dari input pengguna
objek1_1013 = data_1013

# objek2 merujuk pada objek yang sama dengan objek1
objek2_1013 = objek1_1013

# objek3 memiliki isi sama, tetapi merupakan objek baru
objek3_1013 = data_1013.copy()


print("objek1 =", objek1_1013)
print("objek2 =", objek2_1013)
print("objek3 =", objek3_1013)

# Operator is
hasil_1013 = objek1_1013 is objek2_1013
print("\nOperator identitas IS")
print("objek1 is objek2 =", hasil_1013)

# Operator is not
hasil_1013 = objek1_1013 is not objek3_1013
print("\nOperator identitas IS NOT")
print("objek1 is not objek3 =", hasil_1013)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =", objek1_1013 is objek3_1013)
print("objek1 == objek3 =", objek1_1013 == objek3_1013)
```

### 2.2.6 Program "perbandingan_2611531013.py"


```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_1013 = int(input("Input angka-1: "))
angka2_1013 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_1013 = angka1_1013 > angka2_1013
print("\nOperator lebih besar dari")
print("angkal > angka2 =", hasil_1013)

# Lebih kecil dari
hasil_1013 = angka1_1013 < angka2_1013
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =", hasil_1013)

# Lebih besar dari atau sama dengan
hasil_1013 = angka1_1013 >= angka2_1013
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =", hasil_1013)

# Lebih kecil dari atau sama dengan
hasil_1013 = angka1_1013 <= angka2_1013
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =", hasil_1013)

# Sama dengan
hasil_1013 = angka1_1013 == angka2_1013
print("\nOperator sama dengan")
print("angkal == angka2 =", hasil_1013)

# Tidak sama dengan
hasil_1013 = angka1_1013 != angka2_1013
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =", hasil_1013)

# Tambahan: perbandingan berantai dalam Python
hasil_1013 = 0 < angka1_1013 < 100
print("\nPerbandingan berantai")
print("0 < angkal < 100 =", hasil_1013)

hasil_1013 = angka2_1013 < 100
print("0 < angka2 < 100 =", hasil_1013)
```
