## 2.1 Dasar Teori

Perulangan (looping) merupakan struktur dalam pemrograman yang digunakan untuk menjalankan suatu perintah secara berulang berdasarkan kondisi atau jumlah pengulangan tertentu. Dengan perulangan, program dapat menjalankan instruksi yang sama tanpa harus menuliskan kode tersebut berulang kali. Pada Python, perulangan umumnya dilakukan menggunakan `"for"` dan `"while"`.

1. Perulangan dalam Python

    Perulangan merupakan struktur pemrograman yang digunakan untuk menjalankan suatu perintah secara berulang. Python menyediakan beberapa bentuk perulangan, salah satunya adalah for. Perulangan digunakan untuk membuat proses yang sama dapat dilakukan berkali-kali tanpa harus menuliskan perintah yang sama secara berulang. 

2. Perulangan for

    Perulangan for digunakan untuk melakukan iterasi terhadap suatu urutan data atau nilai tertentu. Pada setiap iterasi, variabel akan mengambil nilai dari urutan tersebut dan menjalankan perintah yang berada di dalam blok for.

3. Fungsi range()

    `range()` digunakan untuk menghasilkan urutan bilangan yang dapat digunakan sebagai batas atau jumlah pengulangan pada for. Fungsi ini dapat menentukan nilai awal, nilai akhir, dan interval. Nilai akhir pada range() tidak termasuk dalam urutan yang dihasilkan.

4. Nested for

    Nested for adalah perulangan for yang berada di dalam perulangan for lainnya. Konsep ini digunakan ketika suatu proses membutuhkan lebih dari satu tingkat pengulangan, misalnya untuk memproses baris dan kolom atau membuat pola tertentu.


5. Perulangan for dengan Percabangan

    Perulangan for dapat dikombinasikan dengan if, elif, dan else untuk memberikan kondisi tertentu selama proses pengulangan berlangsung. Dengan kombinasi tersebut, program dapat menentukan perintah yang berbeda berdasarkan kondisi yang terpenuhi pada setiap iterasi.

## 2.2 Kode Program

### 2.2.1 Program "perulangan_for1_2611531013.py"


```python
# Buat file dengan nama perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input() 

ulang_1013 = int(input("Masukkan Jumlah perulangan: "))

for i_1013 in range(ulang_1013):
    print(f"Perulangan ke-{i_1013}")
```

**Penjelasan :**
1. `ulang_1013 = int(input("Masukkan Jumlah perulangan: "))`

- `input()` → meminta input dari pengguna.
- `"Masukkan Jumlah perulangan: "` → pesan yang ditampilkan kepada pengguna.
- `int()` → mengubah input menjadi bilangan bulat.
- `ulang_1013` → variabel untuk menyimpan jumlah perulangan.


2. `for i_1013 in range(ulang_1013):`

- `for` → perintah untuk melakukan perulangan.
- `i_1013` → variabel penghitung perulangan.
- `range(ulang_1013)` → menentukan jumlah perulangan berdasarkan nilai ulang_1013.
- `:` → menandakan awal blok kode perulangan.


3. `print(f"Perulangan ke-{i_1013}")`

- `print()` → menampilkan hasil ke layar.
- `f""` → digunakan untuk memasukkan nilai variabel ke dalam teks.
- `{i_1013}` → menampilkan nilai penghitung pada setiap perulangan.


4. `i_1013` dimulai dari 0

- Karena menggunakan `range()`, perulangan dimulai dari 0. 
- Jika jumlah perulangan yang dimasukkan adalah 5, maka nilai `i_1013` adalah 0, 1, 2, 3, dan 4.


5. Kesimpulan program

- Program meminta jumlah perulangan dari pengguna, kemudian menjalankan perulangan `for` sesuai jumlah tersebut dan menampilkan nomor setiap perulangan.

### 2.2.3 Program "perulangan_for3_2611531013.py"

- Digunakan untuk menambahkan nilai `i_1013` ke dalam `jumlah_1013` pada setiap perulangan.

### 2.2.4 Program "nested_for1_2611531013.py"


```python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_1013 = int(input("masukkan nilai batas: "))
for line_1013 in range(1, batas_1013 + 1):
    for j_1013 in range (1, (-1 * line_1013 + batas_1013) + 1 ):
        print(".", end="")
    print(line_1013)
```

**Penjelasan :**`
1. `batas_1013 = int(input("masukkan nilai batas: "))`
- Digunakan untuk meminta pengguna memasukkan nilai batas perulangan, kemudian disimpan ke variabel `batas_1013` dalam bentuk bilangan bulat.

2. `for line_1013 in range(1, batas_1013 + 1):`
- Digunakan sebagai perulangan luar (outer loop). Perulangan dimulai dari `1` sampai nilai `batas_1013.`

3. `for j_1013 in range(1, (-1 * line_1013 + batas_1013) + 1):`
- Digunakan sebagai perulangan dalam (inner loop). Perulangan ini menentukan berapa banyak tanda titik . yang akan ditampilkan pada setiap baris.

4. `-1 * line_1013`
- Digunakan untuk menghasilkan nilai negatif dari `line_1013.` Semakin besar nilai `line_1013` , jumlah perulangan pada bagian ini akan semakin sedikit.

5. `+ batas_1013`
- Digunakan untuk menyesuaikan jumlah perulangan berdasarkan nilai batas yang dimasukkan pengguna.

6. `print(".", end="")`
- Digunakan untuk menampilkan tanda titik . tanpa berpindah ke baris baru. Karena berada di dalam inner loop, titik akan dicetak berulang kali.

7. `print(line_1013)`
- Digunakan untuk menampilkan nilai `line_1013` setelah titik selesai dicetak, kemudian berpindah ke baris berikutnya.




### 2.2.5 Program "nested_for2_2611531013.py"


```python
# Buat file dengan nama nested_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_1013 = int(input("masukkan nilai batas: "))
for i_1013 in range(1, batas_1013 + 1): 
    for j_1013 in range (1, batas_1013 + 1 ):
        print("*", end="")
    print() # Pindah ke baris berikutnya
```

- Merupakan perulangan luar (outer loop) yang digunakan untuk menentukan jumlah baris. Perulangan berjalan dari 1 sampai `batas_1013.`

3. `for j_1013 in range(1, batas_1013 + 1):`
- Merupakan perulangan dalam (inner loop) yang digunakan untuk menentukan jumlah tanda `*` pada setiap baris.

4. `print("*", end="")`
- Digunakan untuk menampilkan tanda `*. end=""` membuat tanda * tetap berada pada baris yang sama selama inner loop masih berjalan.

5. `print()`
- Digunakan untuk pindah ke baris berikutnya setelah semua tanda `*` pada satu baris selesai dicetak.

### 2.2.7 Program "nested_for4_2611531013.py"


```python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam pytohn
# Nama variabel ditambah 4 digit nim terakhir c_1013ontoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_1013 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1013 % 2 != 0:
    print("Tinggi pola harus bilangan genap.")
else:
    a_1013 = tinggi_1013
    c_1013 = a_1013
    lebar_1013 = (2 * tinggi_1013) - 2

    for i_1013 in range(1, tinggi_1013 + 1):
        b_1013 = c_1013 + 1

        for j_1013 in range(1, lebar_1013 + 1):

            # Baris atas dan bawah
            if i_1013 == 1 or i_1013 == tinggi_1013:
                if j_1013 == 1 or j_1013 == lebar_1013:
                    print("#", end="")
                else:
                    print("*", end="")
            # Baris isi
            else:
                if j_1013 == 1 or j_1013 == lebar_1013:
                    print("|", end="")
                else:
                    if j_1013 == c_1013:
                        print("<", end="")
                    elif j_1013 == b_1013:
                        print(">", end="")
                    elif j_1013 == (lebar_1013 - c_1013):
                        print("<", end="")
                    elif j_1013 == (lebar_1013 - c_1013 + 1):
                        print(">", end="")
                    elif j_1013 > b_1013 and j_1013 < (lebar_1013 - c_1013):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # logika asli java
        a_1013 -= 2
        if a_1013 <= 0:
            c_1013 = -a_1013 + 2
        else:
            c_1013 = a_1013
                    
```

**Penjelasan :**

- Digunakan untuk meminta pengguna memasukkan tinggi pola. Fungsi `int()` digunakan untuk mengubah input menjadi bilangan bulat dan menyimpannya dalam variabel `tinggi_1013.`

- Digunakan untuk memeriksa apakah tinggi pola merupakan bilangan ganjil. Operator `%` digunakan untuk mencari sisa pembagian, sedangkan `!= 0` berarti tidak sama dengan nol.

3. `print("Tinggi pola harus bilangan genap.") dan else:`
- Jika tinggi yang dimasukkan ganjil, program menampilkan pesan bahwa tinggi harus berupa bilangan genap. Jika genap, program melanjutkan proses pembuatan pola melalui `else.`

4. `a_1013 = tinggi_1013` dan `c_1013 = a_1013`
- Digunakan untuk memberikan nilai awal tinggi pola kepada variabel `a_1013` dan `c_1013.`

5. `lebar_1013 = (2 * tinggi_1013) - 2`
- Digunakan untuk menentukan lebar pola berdasarkan tinggi yang dimasukkan.

6. `for i_1013 in range(1, tinggi_1013 + 1):`
- Merupakan perulangan luar yang digunakan untuk mengatur baris pola dari `1` sampai `tinggi_1013.`

7. `b_1013 = c_1013 + 1`
- Digunakan untuk menentukan nilai `b_1013` berdasarkan nilai `c_1013` pada setiap baris.

8. `for j_1013 in range(1, lebar_1013 + 1):`
- Merupakan perulangan dalam yang digunakan untuk mengatur posisi atau kolom pada setiap baris pola.

9. `if i_1013 == 1 or i_1013 == tinggi_1013:`
- Digunakan untuk memeriksa apakah baris yang sedang diproses merupakan baris paling atas atau paling bawah.

10. `if j_1013 == 1 or j_1013 == lebar_1013:`
- Digunakan untuk memeriksa posisi paling kiri atau paling kanan pada baris tersebut. Jika terpenuhi, program menampilkan `#`, sedangkan posisi lainnya pada baris atas dan bawah menampilkan `*`.

11. `else:` pada baris isi
- Digunakan untuk memproses baris yang berada di bagian tengah pola.

12. `if j_1013 == 1 or j_1013 == lebar_1013:`
- Digunakan untuk membuat sisi kiri dan kanan pola dengan menampilkan karakter `|.`

13. `if j_1013 == c_1013` dan `elif j_1013 == b_1013`
- Digunakan untuk menentukan posisi karakter `<` dan `>` berdasarkan `nilai c_1013` dan `b_1013.`

```python
# --- HEADER DAN INPUT PROGRAM ---
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
# Meminta masukan skala N dari pengguna
n_1013 = int(input("Masukkan ukuran skala jam pasir (N): "))

# --- PERULANGAN UTAMA MATRIKS 2D ---
# Loop baris i (bergerak dari atas ke bawah)
for i_1013 in range(1, ((n_1013 * 2) + 4)):
    # Loop kolom j (bergerak dari kiri ke kanan)
    for j_1013 in range(1, ((n_1013 * 4) + 8)):
        
        # --- BINGKAI: PENGECEKAN 4 CORNER / SUDUT TERLUAR ---
        if ((i_1013 == 1 and j_1013 == 1) or (i_1013 == 1 and j_1013 == (n_1013 * 4) + 7) or (i_1013 == (n_1013 * 2) + 3 and j_1013 == (n_1013 * 4) + 7) or (i_1013 == (n_1013 * 2) + 3 and j_1013 == 1)):
            print("#", end="")
            
        else:
            # --- BINGKAI: ATAS, BAWAH, KIRI, DAN KANAN ---
            if (i_1013 == 1 or i_1013 == ((n_1013 * 2) + 3)):
               print("=", end="") # Garis horizontal pembatas atas dan bawah
            elif (j_1013 == 1 or j_1013 == ((n_1013 * 4) + 7)):
                print("|", end="") # Dinding vertikal pembatas kiri dan kanan
                
            else:
                # --- ELEMEN POROS KRISTAL TENGAH (<*>) ---
                if (j_1013 == ((n_1013 * 2) + 3)):
                    print("<", end="") # Karakter kiri poros
                elif (j_1013 == ((n_1013 * 2) + 5)):
                    print(">", end="") # Karakter kanan poros
                elif (j_1013 == ((n_1013 * 2) + 4)):
                    print("*", end="") # Karakter tengah poros
                elif (j_1013 == 2 or j_1013 == (n_1013 * 2) + 6):
                    print(" ", end="") # Spasi pemisah di dekat poros/bingkai
                    
                else:
                    # --- MENENTUKAN SKALA ANGKA AKTIF PER BARIS ---
                    if i_1013 - 1 <= n_1013:
                        baris_1013 = (n_1013 - (i_1013 - 2)) # Fase Atas: Nilai N mengecil ke 1
                    else:
                        baris_1013 = (i_1013 - 2) - n_1013    # Fase Bawah: Nilai 1 membesar ke N

                    # --- PENCETAKAN DERET ANGKA PALINDROM ---
                    # Sisi Kiri Poros: Deret angka menurun (misal: 4 3 2 1)
                    if (j_1013 % 2 == 1) and (((n_1013 * 2) + 3 - (baris_1013 * 2)) <= j_1013 < ((n_1013 * 2) + 3)):
                        print(((n_1013 * 2) + 3 - j_1013) // 2, end="")
                    # Sisi Kanan Poros: Deret angka menaik (misal: 1 2 3 4)
                    elif (j_1013 % 2 == 1) and (((n_1013 * 2) + 5) < j_1013 <= ((n_1013 * 2) + 5 + (baris_1013 * 2))):
                        print((j_1013 - ((n_1013 * 2) + 5)) // 2, end="")
                    # Padding Spasi Kosong di Luar Jangkauan Angka
                    else:
                        print(" ", end="")
                        
    # --- PINDAH KE BARIS BARU ---
    print()
```
